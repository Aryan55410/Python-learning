```python
import pandas  as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

data = pd.read_csv("student-1.csv")

x = data[['hours']]
y = data[['score']]

model = LinearRegression()
model.fit(x, y)

predicted_score = model.predict(x)

#evaluate the model
mae = mean_absolute_error(y, predicted_score)
mse = mean_squared_error(y, predicted_score)
rmse = np.sqrt(mse)

print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)

new_hour = float(input("enter a hour = "))
new_pred = model.predict([[new_hour]])
print(f"prediction for {new_hour} is score = {new_pred[0][0]}")

new_prediction = model.predict([[7]])
print(f"prediction for 7 hour ", new_prediction )
```
